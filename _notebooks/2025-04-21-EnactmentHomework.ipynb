# Selection, Insertion Sort & Big O - Notes

### Key Takeaway  
- Selection Sort: Finds the smallest element and places it in the correct position. Always O(n²).  
- Insertion Sort: Builds a sorted section by inserting elements one at a time. Best case O(n) if nearly sorted.  
- Big O: Describes how runtime or memory scales with input size, not actual speed.

### Visualization  
- Selection: Like picking the smallest card each time and placing it in order.  
- Insertion: Like organizing cards in your hand as you draw them.  
- Big O: Graphs showing how algorithms grow — linear, quadratic, exponential, etc.

### Important Concepts & Vocabulary  
- In-place: Uses no extra memory to sort.  
- Stable: Keeps equal elements in original order (Insertion is stable, Selection is not).  
- Time Complexity:  
  - Selection Sort: O(n²) worst/avg/best  
  - Insertion Sort: Best O(n), Avg/Worst O(n²)  
- Big O Classes:  
  - O(1): Constant  
  - O(log n): Logarithmic  
  - O(n): Linear  
  - O(n²): Quadratic  
  - O(2ⁿ): Exponential  

### Syntax Examples

Selection Sort:
```java  
for (int i = 0; i < arr.length - 1; i++) {  
 int minIndex = i;  
 for (int j = i + 1; j < arr.length; j++) {  
  if (arr[j] < arr[minIndex]) {  
   minIndex = j;  
  }  
 }  
 int temp = arr[i];  
 arr[i] = arr[minIndex];  
 arr[minIndex] = temp;  
}  
```

Insertion Sort: 
```java 
for (int i = 1; i < arr.length; i++) {  
 int key = arr[i];  
 int j = i - 1;  
 while (j >= 0 && arr[j] > key) {  
  arr[j + 1] = arr[j];  
  j--;  
 }  
 arr[j + 1] = key;  
}  
```

### Use Cases  
- Teaching core sorting concepts  
- Insertion is useful for small or nearly sorted datasets  
- When performance is less critical and simplicity is preferred

### Common Mistakes  
- Swapping before finding the min in selection sort  
- Incorrect index updates in insertion sort  
- Misinterpreting Big O as actual runtime  
- Using O(n²) sorts on large datasets without considering efficiency

````java
// Question 1 Part A
public static void selectionSortDescending(int[] arr) {
    for (int i = 0; i < arr.length - 1; i++) {
        int maxIndex = i;
        for (int j = i + 1; j < arr.length; j++) {
            if (arr[j] > arr[maxIndex]) {
                maxIndex = j;
            }
        }
        int temp = arr[i];
        arr[i] = arr[maxIndex];
        arr[maxIndex] = temp;
    }
}

//Part B
public static void insertionSortAscending(int[] arr) {
    for (int i = 1; i < arr.length; i++) {
        int key = arr[i];
        int j = i - 1;
        while (j >= 0 && arr[j] > key) {
            arr[j + 1] = arr[j];
            j--;
        }
        arr[j + 1] = key;
    }
}
````

````java
// Question 2
public class SortingExample {

    // Selection Sort in Descending Order
    public static void selectionSortDescending(int[] arr) {
        for (int i = 0; i < arr.length - 1; i++) {
            int maxIndex = i;
            for (int j = i + 1; j < arr.length; j++) {
                if (arr[j] > arr[maxIndex]) {
                    maxIndex = j;
                }
            }
            int temp = arr[i];
            arr[i] = arr[maxIndex];
            arr[maxIndex] = temp;
        }
    }

    // Insertion Sort in Ascending Order
    public static void insertionSortAscending(int[] arr) {
        for (int i = 1; i < arr.length; i++) {
            int key = arr[i];
            int j = i - 1;
            while (j >= 0 && arr[j] > key) {
                arr[j + 1] = arr[j];
                j--;
            }
            arr[j + 1] = key;
        }
    }

    // Timer for Selection Sort
    public static long timeSelectionSort(int[] arr) {
        int[] copy = arr.clone();  // Clone the array to keep original array intact
        long start = System.nanoTime();
        selectionSortDescending(copy);  // Perform Selection Sort
        long end = System.nanoTime();
        return end - start;  // Return the time taken in nanoseconds
    }

    // Timer for Insertion Sort
    public static long timeInsertionSort(int[] arr) {
        int[] copy = arr.clone();  // Clone the array
        long start = System.nanoTime();
        insertionSortAscending(copy);  // Perform Insertion Sort
        long end = System.nanoTime();
        return end - start;  // Return the time taken in nanoseconds
    }

    // Main method to run the sorting and print the time taken
    public static void main(String[] args) {
        int[] arrayA = {29, 10, 14, 37, 13};  // Array A
        int[] arrayB = {1, 2, 5, 4, 3, 6, 7, 8, 9, 10};  // Array B

        // Measure and print the time taken for each sort on array A
        long selectionTimeA = timeSelectionSort(arrayA);
        long insertionTimeA = timeInsertionSort(arrayA);

        // Measure and print the time taken for each sort on array B
        long selectionTimeB = timeSelectionSort(arrayB);
        long insertionTimeB = timeInsertionSort(arrayB);

        // Output the time results
        System.out.println("Array A:");
        System.out.println("Selection Sort Time: " + selectionTimeA + " ns");
        System.out.println("Insertion Sort Time: " + insertionTimeA + " ns");

        System.out.println("\nArray B:");
        System.out.println("Selection Sort Time: " + selectionTimeB + " ns");
        System.out.println("Insertion Sort Time: " + insertionTimeB + " ns");
    }
}
````

#### Array A: `{29, 10, 14, 37, 13}`
- **Selection Sort**: Fixed number of comparisons, works well for small arrays.
- **Insertion Sort**: Performs better on small, unsorted arrays due to fewer operations.

**Result**: **Insertion Sort** likely performs better on Array A.

#### Array B: `{1, 2, 5, 4, 3, 6, 7, 8, 9, 10}`
- **Selection Sort**: Full scan for max/min, less efficient for nearly sorted arrays.
- **Insertion Sort**: Efficient on nearly sorted data, only a few shifts needed.

**Result**: **Insertion Sort** likely performs better on Array B.

#### When to Use Each Algorithm:
- **Selection Sort**: Best for small datasets and when memory is constrained.
- **Insertion Sort**: Best for nearly sorted or small datasets, stable sort.

**Summary**:
- **Array A**: **Insertion Sort** likely performs better.
- **Array B**: **Insertion Sort** performs better due to near-sorted state.

# Collectables - Notes:

#### 1. **Lists**:
- **Key Concept**: Ordered, index-based collection allowing duplicates.
- **Syntax**: 
  - `ArrayList<Integer> list = new ArrayList<>();`
  - `list.add(10);`
- **Use Cases**: When you need an ordered collection of elements, with random access by index.
- **Common Mistakes**: 
  - Accessing an invalid index.
  - Confusing `LinkedList` (doubly linked) and `ArrayList` (dynamic array) performance characteristics.

#### 2. **Sets**:
- **Key Concept**: Unordered collection of unique elements, no duplicates allowed.
- **Syntax**: 
  - `Set<Integer> set = new HashSet<>();`
  - `set.add(10);`
- **Use Cases**: When you need to store unique elements and order doesn't matter.
- **Common Mistakes**: 
  - Assuming order is preserved (use `LinkedHashSet` for that).
  - Not using `contains()` to check for existence before adding.

#### 3. **Maps**:
- **Key Concept**: Collection of key-value pairs, allows fast retrieval by key.
- **Syntax**:
  - `Map<Integer, String> map = new HashMap<>();`
  - `map.put(1, "one");`
- **Use Cases**: When you need to map keys to values, like in dictionary or database scenarios.
- **Common Mistakes**:
  - Not checking for key existence before calling `get()`.
  - Using an inappropriate implementation like `HashMap` when order matters (use `TreeMap` or `LinkedHashMap`).

#### 4. **Queues**:
- **Key Concept**: FIFO collection, elements are processed in the order they were added.
- **Syntax**:
  - `Queue<Integer> queue = new LinkedList<>();`
  - `queue.offer(10);`
- **Use Cases**: For scheduling tasks, handling requests, or any situation requiring FIFO processing.
- **Common Mistakes**: 
  - Confusing queues with stacks (LIFO).
  - Not using `offer()` (safe method) for adding elements in case of capacity limits.

#### 5. **Deque**:
- **Key Concept**: Double-ended queue allowing elements to be added or removed from both ends.
- **Syntax**:
  - `Deque<Integer> deque = new ArrayDeque<>();`
  - `deque.addFirst(5);`
- **Use Cases**: When you need efficient access to both ends of a collection, such as in implementing stacks or queues.
- **Common Mistakes**: 
  - Using `LinkedList` for deque operations when `ArrayDeque` is more efficient for most use cases.

#### 6. **Stacks**:
- **Key Concept**: LIFO collection, last element added is the first one removed.
- **Syntax**: 
  - `Stack<Integer> stack = new Stack<>();`
  - `stack.push(10);`
- **Use Cases**: When you need to process items in reverse order, such as in recursive algorithms or undo functionality.
- **Common Mistakes**: 
  - Using `Stack` which is slower than `ArrayDeque` in many cases.
  - Not checking if the stack is empty before calling `pop()` or `peek()`.

### Visualization:
- **List**: `1 → 2 → 3 → 4` (Access via index)
- **Set**: `{1, 2, 3, 4}` (No duplicates, unordered)
- **Map**: `{1 -> "One", 2 -> "Two", 3 -> "Three"}` (Key-Value pairs)
- **Queue**: `front → 1 → 2 → 3 → 4 ← rear` (FIFO)
- **Deque**: `front → 1 → 2 → 3 → 4 ← rear` (Add/Remove from both ends)
- **Stack**: `top → 4 → 3 → 2 → 1` (LIFO)

````java
// Question 2
import java.util.*;

public class CommonElements {

    // Method to get common elements between two sets
    public static Set<Integer> getCommonElements(Set<Integer> set1, Set<Integer> set2) {
        Set<Integer> commonSet = new HashSet<>();
        for (Integer element : set1) {
            if (set2.contains(element)) {
                commonSet.add(element);
            }
        }
        return commonSet;
    }

    public static void main(String[] args) {
        // Test the method with sample sets
        Set<Integer> set1 = new HashSet<>(Arrays.asList(1, 2, 3, 4, 5));
        Set<Integer> set2 = new HashSet<>(Arrays.asList(4, 5, 6, 7));
        Set<Integer> common = getCommonElements(set1, set2);
        System.out.println("Common elements: " + common);
    }
}
````

# Merge Sort - Notes

### Key Takeaway
- Divide-and-conquer sorting algorithm
- Splits array recursively, then merges sorted halves
- Consistent **O(n log n)** time complexity in all cases

### Visualization
- Recursive breakdown:
  - `[8, 3, 5, 4]` → `[8, 3]`, `[5, 4]`  
  - → `[8] [3] [5] [4]`  
  - → `[3, 8] [4, 5]`  
  - → `[3, 4, 5, 8]`

### Concepts, Vocab & Syntax
- `mergeSort(arr, left, right)`
- `merge(arr, left, mid, right)`
- **Base case**: one-element array
- Combine sorted arrays using two-pointer merge

### Use Cases
- Large data sets
- Requires a stable sort
- Linked list sorting (easy merging)

### Common Mistakes
- Skipping the midpoint in merge
- Forgetting to copy remaining elements
- Overusing temp arrays or copying inefficiently

Merge Sort is a divide-and-conquer algorithm that splits an array into halves, sorts each half, and merges them back together. 
It starts by dividing the array until each subarray has one element, then merges these in sorted order.
This process is repeated recursively, resulting in a fully sorted array.

The time complexity is O(n log n) because the array is divided log n times (halving each time), and each merge step takes O(n) time to combine elements.
Unlike Bubble Sort, which repeatedly swaps adjacent elements and has O(n^2) time complexity, Merge Sort is much more efficient for large datasets.
Compared to Quick Sort, Merge Sort has a more predictable performance (always O(n log n)), but it requires additional memory for merging.

````java
// Question 2: Option B
public class MergeSortInversions {

    public static int countInversions(int[] arr) {
        if (arr == null || arr.length < 2) return 0;
        return mergeSortAndCount(arr, 0, arr.length - 1);
    }

    private static int mergeSortAndCount(int[] arr, int left, int right) {
        if (left >= right) return 0;
        int mid = (left + right) / 2;
        int count = 0;
        count += mergeSortAndCount(arr, left, mid);
        count += mergeSortAndCount(arr, mid + 1, right);
        count += mergeAndCount(arr, left, mid, right);
        return count;
    }

    private static int mergeAndCount(int[] arr, int left, int mid, int right) {
        int[] leftArr = java.util.Arrays.copyOfRange(arr, left, mid + 1);
        int[] rightArr = java.util.Arrays.copyOfRange(arr, mid + 1, right + 1);

        int i = 0, j = 0, k = left, swaps = 0;

        while (i < leftArr.length && j < rightArr.length) {
            if (leftArr[i] <= rightArr[j]) {
                arr[k++] = leftArr[i++];
            } else {
                arr[k++] = rightArr[j++];
                swaps += (mid + 1) - (left + i); // Count inversions
            }
        }

        while (i < leftArr.length) {
            arr[k++] = leftArr[i++];
        }

        while (j < rightArr.length) {
            arr[k++] = rightArr[j++];
        }

        return swaps;
    }
}
````
