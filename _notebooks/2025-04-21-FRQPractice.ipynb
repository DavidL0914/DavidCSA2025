### Question 1

```java
// Part A
private String recombine(String word1, String word2) {
    int half1 = word1.length() / 2;
    int half2 = word2.length() / 2;
    return word1.substring(0, half1) + word2.substring(half2);
}

//Part B
private String[] mixedWords(String[] words) {
    String[] result = new String[words.length];
    for (int i = 0; i < words.length; i += 2) {
        result[i] = recombine(words[i], words[i + 1]);
        result[i + 1] = recombine(words[i + 1], words[i]);
    }
    return result;
}
```

### Things to Look Out For

- Ensure averaging uses double precision.
- Only read from the array — no in-place updates.

### Tips
- Return the original temperature if the cell is on the border.
- For interior cells, calculate the average of the 4 directly adjacent neighbors (up, down, left, right).

### Question 2

```java
// Part A
public static int getPeakIndex(int[] array) {
    for (int i = 1; i < array.length - 1; i++) {
        if (array[i] > array[i - 1] && array[i] > array[i + 1]) {
            return i; // Found peak
        }
    }
    return -1; // No peak found
}


//Part B
public static boolean isMountain(int[] array) {
    int peakIndex = getPeakIndex(array);

    if (peakIndex == -1) {
        return false; // No peak
    }

    // Check increasing before peak
    for (int i = 0; i < peakIndex; i++) {
        if (array[i] >= array[i + 1]) {
            return false;
        }
    }

    // Check decreasing after peak
    for (int i = peakIndex; i < array.length - 1; i++) {
        if (array[i] <= array[i + 1]) {
            return false;
        }
    }

    return true;
}
```

## Key Takeaways

### Common Mistakes
- Modifying temps before finishing all tolerance checks.
- Comparing new values to the wrong original ones.
- Forgetting to update temps after successful validation.

### What I Learned
- Separating computation and mutation makes logic cleaner and safer.
- Use absolute difference to enforce tolerance checks.

### Things to Look Out For
- Correct usage: Math.abs(new - old) > tolerance.
- Keep array dimensions consistent when copying.
- Don’t skip any cell while iterating and updating.

### Tips
- Use a temporary 2D array to store computed temperatures.
- Only update the original temps array after confirming all values are within tolerance.
- Compare each new value with the original using Math.abs.

```java
// Part A
private double computeTemp(int row, int col) {
    if (row == 0 || col == 0 || row == temps.length - 1 || col == temps[0].length - 1) {
        return temps[row][col];
    }

    double sum = temps[row - 1][col] + temps[row + 1][col] +
                 temps[row][col - 1] + temps[row][col + 1];

    return sum / 4.0;
}


//Part B
public boolean updateAllTemps(double tolerance) {
    double[][] updatedTemps = new double[temps.length][temps[0].length];
    boolean allWithinTolerance = true;

    for (int row = 0; row < temps.length; row++) {
        for (int col = 0; col < temps[0].length; col++) {
            updatedTemps[row][col] = computeTemp(row, col);
            double diff = Math.abs(updatedTemps[row][col] - temps[row][col]);
            if (diff > tolerance) {
                allWithinTolerance = false;
            }
        }
    }

    // Replace temps with updated values
    for (int row = 0; row < temps.length; row++) {
        for (int col = 0; col < temps[0].length; col++) {
            temps[row][col] = updatedTemps[row][col];
        }
    }

    return allWithinTolerance;
}
```

```java
//Part A
public boolean record(int score) {
    for (int i = 0; i < scoreList.size(); i++) {
        ScoreInfo info = scoreList.get(i);
        if (info.getScore() == score) {
            info.increment();
            return false;
        } else if (info.getScore() > score) {
            scoreList.add(i, new ScoreInfo(score));
            return true;
        }
    }
    scoreList.add(new ScoreInfo(score)); // If score is largest, add at end
    return true;
}

//Part B
public void recordScores(int[] stuScores) {
    for (int score : stuScores) {
        record(score);
    }
}
```

## Key Takeaways

### Common Mistakes
- Forgetting to maintain **increasing order** in `scoreList`.
- Not checking for **existing scores** before adding a new one.
- Using `==` for object comparison instead of proper field access (e.g., `getScore()`).
- Adding scores without returning the correct `true`/`false` value in `record`.

### What I Learned
- How to maintain a sorted list while inserting elements.
- How to avoid duplicates using custom class checks.
- Using `ArrayList` operations (`add(index, element)`) to keep order.
- How frequency tracking works through object methods like `increment()`.

### Things to Look Out For
- Edge case: inserting at the **end** of the list.
- Edge case: list is **empty** initially.
- Don't modify the list while looping with `for-each` — use index-based loop.

### Tips
- Always test with a mix of new and duplicate scores.
- Think through how the loop will behave for each case (less than, equal to, greater than).
- Use helper methods like `increment()` to keep logic clean.

### Question 1

```java
// Part A
public static int hailstoneLength(int n) {
    int count = 1; // Include the starting number
    while (n != 1) {
        if (n % 2 == 0) {
            n = n / 2;
        } else {
            n = 3 * n + 1;
        }
        count++;
    }
    return count;
}

// Part B
public static boolean isLongSequence(int n) {
    return hailstoneLength(n) > n;
}

// Part C
public static double propLong(int n) {
    int longCount = 0;
    for (int i = 1; i <= n; i++) {
        if (isLongSequence(i)) {
            longCount++;
        }
    }
    return (double) longCount / n;
}
```

### Question 2

```java
import java.util.Random;

public class GameSpinner {
    private int sectors;
    private int lastSpin;
    private int runLength;
    private Random rand;

    // Constructor
    public GameSpinner(int numSectors) {
        sectors = numSectors;
        lastSpin = -1; // No spin has occurred yet
        runLength = 0;
        rand = new Random();
    }

    // Spins the spinner and returns a random sector value
    public int spin() {
        int spinResult = rand.nextInt(sectors) + 1; // random from 1 to sectors

        if (spinResult == lastSpin) {
            runLength++;
        } else {
            runLength = 1;
            lastSpin = spinResult;
        }

        return spinResult;
    }

    // Returns the current run length of the same result
    public int currentRun() {
        return runLength;
    }
}
```

```java
// Part A
public Theater(int seatsPerRow, int tier1Rows, int tier2Rows)
{
    theaterSeats = new Seat[tier1Rows + tier2Rows][seatsPerRow];
    
    for (int row = 0; row < theaterSeats.length; row++)
    {
        int tier = (row < tier1Rows) ? 1 : 2;
        
        for (int col = 0; col < seatsPerRow; col++)
        {
            theaterSeats[row][col] = new Seat(true, tier);
        }
    }
}

// Part B
public boolean reassignSeat(int fromRow, int fromCol, int toRow, int toCol)
{
    Seat fromSeat = theaterSeats[fromRow][fromCol];
    Seat toSeat = theaterSeats[toRow][toCol];
    
    if (toSeat.isAvailable() && toSeat.getTier() >= fromSeat.getTier())
    {
        fromSeat.setAvailability(true);  // vacate source seat
        toSeat.setAvailability(false);   // occupy destination seat
        return true;
    }
    
    return false;
}
```
